# Example Sentences

**Das ist mein Vater.**  
This is my father.

**Das ist meine Mutter.**  
This is my mother.

**Das ist mein Bruder.**  
This is my brother.

**Das ist meine Schwester.**  
This is my sister.

# Talking About Your Family

**Das ist meine Familie.**  
This is my family.

**Mein Vater heißt Peter.**  
My father's name is Peter.

**Meine Mutter heißt Anna.**  
My mother's name is Anna.

**Ich habe einen Bruder.**  
I have a brother.

**Ich habe eine Schwester.**  
I have a sister.

# Talking About Family Size

You can describe **how big or small your family is** in German.

### Useful Words

| German | English |
|------|------|
| groß | big |
| klein | small |
| die Person | person |
| die Personen | people |
| viele | many |
| wenige | few |

---

# Example Sentences

**Meine Familie ist groß.**  
My family is big.

**Meine Familie ist klein.**  
My family is small.

**Wir sind vier Personen.**  
We are four people.

**Wir sind eine kleine Familie.**  
We are a small family.

**Wir sind eine große Familie.**  
We are a big family.

---

# Talking About Family Members

**Ich habe zwei Brüder.**  
I have two brothers.

**Ich habe eine Schwester.**  
I have one sister.

**Ich habe keine Geschwister.**  
I have no siblings.

**Wir sind fünf Personen in meiner Familie.**  
There are five people in my family.

---

# Example Description

**Meine Familie ist klein.  
Wir sind vier Personen.  
Ich habe einen Bruder und eine Schwester.**

My family is small.  
We are four people.  
I have one brother and one sister.
