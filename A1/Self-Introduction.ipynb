# German Self-Introduction Phrases

This list includes common German phrases for introducing yourself, with their English meanings.

| **German** | **English** |
|------------|------------|
| **Ich heiße Monika. / Ich bin Monika. / Mein Name ist Monika.** | My name is Monika. / I am Monika. |
| **Mein Familienname ist Schmidt.** | My last name is Schmidt. |
| **Ich komme aus Deutschland.** | I come from Germany. |
| **Ich wohne in Berlin.** | I live in Berlin. |
| **Ich spreche Deutsch und Englisch.** | I speak German and English. |
| **Ich bin 23 Jahre alt.** | I am 23 years old. |
| **Ich bin Student (Studentin).** | I am a student (female: Studentin). |
| **Ich bin Lehrer / Lehrerin.** | I am a teacher (female: Lehrerin). |
| **Ich bin ledig / verheiratet.** | I am single / married. |
| **Ich habe ein Kind.** | I have one child. |
| **Ich habe zwei / drei / keine Kinder.** | I have two / three / no children. |
| **Meine Hobbys sind fernsehen und Musik hören.** | My hobbies are watching TV and listening to music. |

# German Self-Introduction – Formal and Informal Questions

This table includes both formal (**Sie**) and informal (**du**) questions with example answers and English meanings.

| **Formal (Sie)** | **Informal (du)** | **Example Answer** | **English Meaning** |
|------------------|-------------------|--------------------|--------------------|
| Wie heißen Sie? | Wie heißt du? | Ich heiße Monika. | What is your name? — My name is Monika. |
| Wie ist Ihr Familienname? | Wie ist dein Familienname? | Mein Familienname ist Schmidt. | What is your last name? — My last name is Schmidt. |
| Woher kommen Sie? | Woher kommst du? | Ich komme aus Deutschland. | Where are you from? — I come from Germany. |
| Wo wohnen Sie? | Wo wohnst du? | Ich wohne in Berlin. | Where do you live? — I live in Berlin. |
| Wie alt sind Sie? | Wie alt bist du? | Ich bin 23 Jahre alt. | How old are you? — I am 23 years old. |
| Welche Sprachen sprechen Sie? | Welche Sprachen sprichst du? | Ich spreche Deutsch und Englisch. | Which languages do you speak? — I speak German and English. |
| Was machen Sie beruflich? | Was machst du beruflich? | Ich bin Lehrerin. | What do you do for a living? — I am a teacher. |
| Was sind Ihre Hobbys? | Was sind deine Hobbys? | Meine Hobbys sind fernsehen und Musik hören. | What are your hobbies? — My hobbies are watching TV and listening to music. |
| Sind Sie verheiratet? | Bist du verheiratet? | Ja, ich bin verheiratet. | Are you married? — Yes, I am married. |
| Haben Sie Kinder? | Hast du Kinder? | Ja, ich habe zwei Kinder. | Do you have children? — Yes, I have two children. |
