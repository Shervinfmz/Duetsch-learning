# 2️⃣1️⃣ Negative Articles in the Nominative (Negationsartikel)

## What is a Negative Article?

The negative article **kein / keine** is used to negate nouns.

It replaces the **indefinite article (ein / eine)**.

Examples:

- Das ist **kein Apfel**. Das ist eine Birne.  
- Das ist **keine Tasche**. Das ist ein Buch.

---

# Asking and Answering Questions

### Example 1

**Ist das ein Kuli?**  
Is this a pen?

**Nein, das ist kein Kuli.**  
Das ist ein Bleistift.

---

### Example 2

**Ist das eine Blume?**  
Is this a flower?

**Nein, das ist keine Blume.**  
Das ist eine Schokolade.

---

### Example 3

**Ist das ein Buch?**  
Is this a book?

**Nein, das ist kein Buch.**  
Das ist ein Handy.

---

### Example 4 (Plural)

**Sind das Sterne?**  
Are those stars?

**Nein, das sind keine Sterne.**  
Das sind Ballons.

# Quick Comparison

| | Indefinite Article | Negative Article |
|---|---|---|
| maskulin | ein | kein |
| feminin | eine | keine |
| neutrum | ein | kein |
| plural | — | keine |

---

# Key Rule

✔ **kein / keine** negates nouns  
✔ It behaves like **ein / eine**  
✔ **Plural always uses keine**

Examples:

- Das ist **kein** Apfel.  
- Das ist **keine** Blume.  
- Das sind **keine** Sterne.
