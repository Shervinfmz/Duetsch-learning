# 2️⃣6️⃣ Der Akkusativ (Accusative Case)

---

# Learning Objectives

- What is **Akkusativ**? (Questions **Wen?** and **Was?**)  
- Definite articles in **Akkusativ**  
- Indefinite articles in **Akkusativ**  
- Negative articles in **Akkusativ**

---

# What is Akkusativ?

The **Akkusativ** marks the **direct object** in a sentence.

The object answers:

| Question | Meaning |
|---|---|
| Wen? | whom |
| Was? | what |

Example:

**Sabine hat einen Bruder.**  
Sabine has one brother.

- **Sabine** → Subject (Nominativ)  
- **einen Bruder** → Object (Akkusativ)

Another example:

**Wir haben ein Kind.**  
We have one child.

---

# Object Types

| Type | Example |
|---|---|
| Person | Petra liebt ihren Mann. |
| Thing | Ich habe ein Auto. |

---

# Indefinite Articles in Akkusativ

| Gender | Nominativ | Akkusativ |
|---|---|---|
| Maskulin | ein | **einen** |
| Feminin | eine | eine |
| Neutrum | ein | ein |
| Plural | – | – |

---

# Examples

### Masculine

**Ich esse einen Apfel.**  
I eat an apple.

**Ich habe einen Bruder.**  
I have a brother.

---

### Feminine

**Ich habe eine Katze.**  
I have a cat.

**Ich esse eine Pizza.**

---

### Neuter

**Wir haben ein Baby.**

**Ich esse ein Sandwich.**

---

### Plural

**Ich habe zwei Brüder.**

Plural nouns **do not use ein/eine**.

---

# Definite Articles in Akkusativ

| Gender | Nominativ | Akkusativ |
|---|---|---|
| Maskulin | der | **den** |
| Feminin | die | die |
| Neutrum | das | das |
| Plural | die | die |

---

# Examples

### Masculine

**Wie findest du den Hut?**  
How do you like the hat?

**Den Hut finde ich schön.**

---

### Feminine

**Wie findest du die Hose?**

**Die Hose finde ich cool.**

---

### Neuter

**Wie findest du das Kleid?**

**Das Kleid finde ich elegant.**

---

### Plural

**Wie findest du die Schuhe?**

**Die Schuhe finde ich teuer.**

---

# Negative Articles in Akkusativ

Negative articles (**kein / keine / keinen**) are used to say **not a / no**.

| Gender | Nominativ | Akkusativ |
|---|---|---|
| Maskulin | kein | **keinen** |
| Feminin | keine | keine |
| Neutrum | kein | kein |
| Plural | keine | keine |

---

# Examples

**Hast du einen Kuli?**  
Do you have a pen?

**Nein, ich habe keinen Kuli.**

---

**Hast du eine Uhr?**  
Do you have a watch?

**Nein, ich habe keine Uhr.**

---

**Hast du ein Lineal?**

**Nein, ich habe kein Lineal.**

---

# Articles in Akkusativ Overview

| Gender | Definite | Indefinite | Negative |
|---|---|---|---|
| Maskulin | **den** | **einen** | **keinen** |
| Feminin | die | eine | keine |
| Neutrum | das | ein | kein |
| Plural | die | – | keine |

---

# Important Rule

Only **masculine articles change in Akkusativ**.

Examples:

**der Mann → den Mann**  
Ich sehe **den Mann**.

**ein Mann → einen Mann**  
Ich habe **einen Mann** gesehen.

**kein Mann → keinen Mann**  
Ich sehe **keinen Mann**.

---

# Quick Visual Rule

| Article Type | Masculine Change |
|---|---|
| der | **den** |
| ein | **einen** |
| kein | **keinen** |

Feminin, Neutrum, and Plural **stay the same**.

---

# Example Sentences

**Ich habe einen Bruder.**  
I have a brother.

**Ich sehe den Mann.**  
I see the man.

**Ich habe keinen Kuli.**  
I don't have a pen.

**Ich habe keine Uhr.**  
I don't have a watch.
