# Possessive Pronouns Overview

| Pronoun | Masculine | Feminine | Neuter | Plural |
|------|------|------|------|------|
| ich | mein | meine | mein | meine |
| du | dein | deine | dein | deine |
| er / es | sein | seine | sein | seine |
| sie (sing.) | ihr | ihre | ihr | ihre |
| wir | unser | unsere | unser | unsere |
| ihr | euer | eure | euer | eure |
| Sie | Ihr | Ihre | Ihr | Ihre |
| sie (plural) | ihr | ihre | ihr | ihre |

---

# Basic Rule

The possessive pronoun changes depending on the **gender of the noun**.

| Article | Possessive Form |
|------|------|
| der | mein |
| das | mein |
| die | meine |
| die (Plural) | meine |

Examples:

- **der Fernseher → mein Fernseher**  
- **das Auto → mein Auto**  
- **die Katze → meine Katze**  
- **die Bücher → meine Bücher**

---

# Examples with "ich"

### der Fernseher (masculine)

Das ist **mein Fernseher**.  
This is my television.

### das Auto (neuter)

Das ist **mein Auto**.  
This is my car.

### die Katze (feminine)

Das ist **meine Katze**.  
This is my cat.

### die Bücher (plural)

Das sind **meine Bücher**.  
These are my books.
