# 🔟 Introducing People (Jemanden vorstellen)

## Important Question

| German | English |
|--------|----------|
| Wer ist das? | Who is this? |

---

# 👩 Example 1 – Sofia

Das ist Sofia.  
Sie kommt aus der Slowakei.  
Sie wohnt in Berlin.  
Sie ist Managerin von Beruf.  
Sie spricht Deutsch, Englisch und Slowakisch.

---

# 👨 Example 2 – Peter

### Questions

- Wer ist das?  
- Woher kommt er?  
- Wo wohnt er?  
- Wie alt ist er?  
- Was lernt er?  
- Wo arbeitet er?  
- Was ist er von Beruf?  
- Welche Sprachen spricht er?  
- Was sind seine Hobbys?  

### Answers

Das ist Peter.  
Er kommt aus Spanien.  
Er wohnt in Frankfurt.  
Er ist 30 Jahre alt.  
Er lernt Deutsch.  
Er arbeitet bei Siemens.  
Er ist Analyst.  
Er spricht Spanisch und Englisch.  
Seine Hobbys sind Musik hören und Lesen.

# Grammar Focus

✔ er → masculine  
✔ sie → feminine  
✔ es → neuter  
✔ sie → plural  

✔ kommt / wohnt / spricht → 3rd person singular  
✔ kommen / wohnen / sprechen → plural
