# 2️⃣2️⃣ Time (Die Zeit)

---

# Basic Time Vocabulary

| German | English |
|------|------|
| die Zeit | time |
| die Uhr | clock |
| die Uhren | clocks |

---

# Units of Time

| German | Plural | English |
|------|------|------|
| die Woche | die Wochen | week / weeks |
| der Tag | die Tage | day / days |
| die Stunde | die Stunden | hour / hours |
| die Minute | die Minuten | minute / minutes |
| die Sekunde | die Sekunden | second / seconds |

Examples:

- Eine **Woche** hat **7 Tage**.  
- Ein **Tag** hat **24 Stunden**.  
- Eine **Stunde** hat **60 Minuten**.  
- Eine **Minute** hat **60 Sekunden**.

---

# Two Ways of Saying Time

| German | Meaning |
|------|------|
| **inoffizielle Zeit** | 12-hour format |
| **offizielle Zeit** | 24-hour format |

Examples:

12-hour format → everyday speech  
24-hour format → official schedules, trains, airports

---

# Asking for the Time

Two common questions:

**Wie spät ist es?**  
What time is it?

**Wie viel Uhr ist es?**  
What is the time?

---

# Saying the Time

Structure:

**Es ist + time + Uhr**

Examples:

| Time | German |
|------|------|
| 01:00 | Es ist **ein Uhr**. |
| 06:00 | Es ist **sechs Uhr**. |
| 15:00 | Es ist **fünfzehn Uhr**. |
| 00:00 | Es ist **null Uhr**. |

---

# Minutes

| Time | German |
|------|------|
| 01:15 | Es ist **ein Uhr fünfzehn**. |
| 01:30 | Es ist **ein Uhr dreißig**. |
| 01:45 | Es ist **ein Uhr fünfundvierzig**. |

---

# Important Rule

When telling time in German:

**Es ist + number + Uhr**

Example:

- Es ist **zwei Uhr**.  
- Es ist **zwölf Uhr**.

# Informal Time in German

**Inoffizielle Zeit**  
Unofficial / everyday spoken time

Also called:

**Zeit in Umgangssprache**  
Time in colloquial language

---

# Asking the Time

**Wie spät ist es?**  
What time is it?

---

# Full Hour

| Time | German |
|-----|------|
| 01:00 / 13:00 | Es ist **eins**. |
| 06:00 / 18:00 | Es ist **sechs**. |
| 12:00 | Es ist **zwölf**. |
| 00:00 | Es ist **Mitternacht**. |

---

# After the Hour (nach)

| Time | German | Meaning |
|-----|------|------|
| 01:10 | Es ist **zehn nach eins** | 10 past 1 |
| 01:15 | Es ist **Viertel nach eins** | quarter past 1 |
| 01:20 | Es ist **zwanzig nach eins** | 20 past 1 |

---

# Before the Next Hour (vor)

| Time | German | Meaning |
|-----|------|------|
| 01:40 | Es ist **zwanzig vor zwei** | 20 to 2 |
| 01:45 | Es ist **Viertel vor zwei** | quarter to 2 |
| 01:50 | Es ist **zehn vor zwei** | 10 to 2 |
| 01:55 | Es ist **fünf vor zwei** | 5 to 2 |

---

# Regional Style (Germany / Austria)

| Time | German |
|-----|------|
| 13:15 | Es ist **Viertel zwei** |
| 13:45 | Es ist **Dreiviertel zwei** |

---

# Half Hour

| Time | German | Meaning |
|-----|------|------|
| 01:30 | Es ist **halb zwei** | half to two (1:30) |

⚠️ Important:  
German **halb zwei = 1:30** (half to two)

---

# Around Half

| Time | German |
|-----|------|
| 01:25 | Es ist **fünf vor halb zwei** |
| 01:35 | Es ist **fünf nach halb zwei** |

---

# Approximate Time

| Time | German | Meaning |
|-----|------|------|
| 04:58 | Es ist **kurz vor fünf** | just before 5 |
| 04:58 | Es ist **fast fünf** | almost 5 |
| 09:34 | Es ist **kurz nach halb zehn** | shortly after 9:30 |
