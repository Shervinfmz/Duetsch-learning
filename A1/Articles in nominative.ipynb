# 1️⃣1️⃣ Articles in the Nominative (Teil 1)

## Learning Objectives

- What is **Nominativ**? (answers the questions: *Wer?* / *Was?*)
- What is a **Substantiv** (noun)?
- Types of articles
- Definite article (bestimmter Artikel)

---

# What is Nominative?

✔ The **Nominative** is the **subject** of the sentence.  
✔ The subject answers: **Wer? (Who?)** or **Was? (What?)**

### Sentence Structure

Satz = **Subjekt (Nominativ) + Verb**

Example:

- Sabine ist ledig.  
- Das Haus ist groß.

---

# Subject (Subjekt)

The subject can be:

- **Nomen** (noun)
- **Pronomen** (pronoun)

---

# Nomen (Substantiv)

A noun can be:

- **Person**
- **Sache** (thing)

Examples:

### Person
- Wer ist das?  
  → Das ist Peter.

### Sache
- Was ist das?  
  → Das ist ein Bus.

---

# Types of Articles

- bestimmter Artikel (definite article)
- unbestimmter Artikel (indefinite article)
- Negationsartikel (negative article)

---

# Definite Articles in Nominative

## Singular

| Gender      | Article |
|-------------|----------|
| maskulin    | der |
| feminin     | die |
| neutrum     | das |

## Plural (all genders)

| Plural |
|--------|
| die |

---

# Examples

| Singular | Plural |
|----------|----------|
| der Mann | die Männer |
| die Frau | die Frauen |
| das Baby | die Babys |

---

# Quick Recognition Trick

Often the noun endings help:

- **der Mann**
- **die Frau**
- **das Baby**

But ⚠️ always learn nouns with their article.

# 1️⃣2️⃣ Indefinite Articles in the Nominative  
(unbestimmte Artikel im Nominativ)

---

# What is an Indefinite Article?

Used when we talk about something **for the first time**  
or when it is **not specific**.

English: a / an  
German: ein / eine

---

# Indefinite Articles – Singular Only

| Gender   | Article |
|----------|----------|
| maskulin | ein |
| feminin  | eine |
| neutrum  | ein |

⚠️ There is **no indefinite article in plural** in German.

---

# Examples

### 🍎 Neutrum

- Das ist **ein Apfel**.  
- Der Apfel ist süß.

---

### 👨 Maskulin

- Das ist **ein Mann**.  
- Der Mann wohnt in Paris.

---

### 👩 Feminin

- Das ist **eine Frau**.  
- Die Frau hört Musik.

---

### 👧 Neutrum (important!)

- Das ist **ein Mädchen**.  
- Das Mädchen tanzt.

⚠️ Even though *Mädchen* means “girl”, it is **neuter** → das

---

# Quick Comparison

| Indefinite | Definite |
|------------|-----------|
| ein Mann   | der Mann |
| eine Frau  | die Frau |
| ein Baby   | das Baby |

---

# Rule Summary

✔ ein → masculine & neuter  
✔ eine → feminine  
✔ No plural form  
✔ Use for something **not specific** or **first mention**
