# German – Asking "How are you?"

## Asking How Someone Is

| **German** | **English** |
|------------|------------|
| Wie geht's? | How are you? |
| Wie geht es Ihnen? | How are you? (formal) |
| Wie geht es dir? | How are you? (informal) |
| Wie geht’s? | How are you? (informal, short form) |

---

## Positive Responses

| **German** | **English** |
|------------|------------|
| Mir geht es gut! | I am fine. |
| Mir geht es sehr gut! | I am very well. |
| super! | super! |
| wunderbar! | wonderful! |
| ausgezeichnet! | excellent! |
| toll! | awesome! |
| prima! | great! |
| fantastisch! | fantastic! |

---

## Neutral Response

| **German** | **English** |
|------------|------------|
| Es geht! | Not too bad. |

---

## Negative Responses

| **German** | **English** |
|------------|------------|
| Mir geht es schlecht! | I am not doing well. |
| Mir geht es nicht gut! | I am not so well. |
| Mir geht es sehr schlecht! | I am feeling very bad. |
| miserabel! | miserable! |
| furchtbar! | awful! |
| Ich bin müde. | I am tired. |
| Ich bin krank. | I am unwell. |

---

## Asking Back

| **German** | **English** |
|------------|------------|
| Und Ihnen? | And you? (formal) |
| Und dir? | And you? (informal) |
| Auch gut, danke! | Also good, thanks! |
