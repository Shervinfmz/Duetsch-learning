# German Sentence Structure & Verb Conjugation (At a Glance)

---

## Basic Sentence Rules

- ✔ The **subject conjugates the verb**.
- ✔ In normal statements, the **verb is in the second position**.

### Example (Statement)

Subject + Verb + ...

- Ich komme aus Deutschland.
- Du wohnst in Berlin.
- Sie sprechen Deutsch.

---

## Question Types

### 1️⃣ W-Questions (W-Fragen)

Questions beginning with **W-words** (Wo? Woher? Wie? Was? Warum? etc.)

➡ The **verb is in the second position**.

**Structure:**

W-word + Verb + Subject + ...

**Examples:**

- Wie heißen Sie?
- Wo wohnst du?
- Woher kommen Sie?
- Welche Sprachen sprichst du?

---

### 2️⃣ Yes/No Questions (Ja/Nein-Fragen)

Questions answered with **"ja"** or **"nein"**

➡ The **verb is in the first position**.

**Structure:**

Verb + Subject + ...

**Examples:**

- Bist du verheiratet?
- Haben Sie Kinder?
- Kommst du aus Deutschland?
- Sprechen Sie Deutsch?

---

## Personal Pronouns

There are various pronouns such as:

- **ich** (I)
- **du** (you – informal singular)
- **Sie** (you – formal singular)

---

## Full Personal Pronoun Overview

| Person | Singular | Plural |
|--------|----------|--------|
| **1st person** | ich | wir |
| **2nd person** | du | ihr (informal) |
|  | Sie (formal) | Sie (formal) |
| **3rd person** | er (masculine) | sie |
|  | sie (feminine) |  |
|  | es (neuter) |  |

---

## Verb Endings (Present Tense – Regular Verbs)

- Verb ending for **"ich"** is **-e**
- Verb ending for **"du"** is **-st**
- Verb ending for **"Sie"** is **-en**

### Example with "kommen" (to come)

| Pronoun | Conjugation |
|---------|------------|
| ich     | ich komme  |
| du      | du kommst  |
| Sie     | Sie kommen |

---

# Irregular Verb: **haben** (to have)

## Conjugation of *haben*

| Person | Singular | Plural |
|--------|----------|--------|
| ich    | habe     | wir haben |
| du     | hast     | ihr habt |
| er/sie/es | hat  | sie haben |
| Sie (formal) | haben | Sie haben |

### Examples

- Ich habe zwei Kinder.
- Du hast Zeit.
- Er hat ein Auto.
- Wir haben Deutschunterricht.
- Sie haben ein Haus.

---

# Irregular Verb: **sein** (to be)

## Conjugation of *sein*

| Person | Singular | Plural |
|--------|----------|--------|
| ich    | bin      | wir sind |
| du     | bist     | ihr seid |
| er/sie/es | ist  | sie sind |
| Sie (formal) | sind | Sie sind |

### Examples

- Ich bin 23 Jahre alt.
- Du bist Student.
- Er ist Lehrer.
- Wir sind müde.
- Sie sind verheiratet.

---

## Important Notes

- The **subject determines the verb ending**.
- The verb position changes depending on the sentence type.
- Some verbs (like *haben* and *sein*) are irregular.
- There are additional exceptions which you will learn later.
