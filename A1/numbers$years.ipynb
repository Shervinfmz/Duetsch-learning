# 8️⃣ Dates, Years & Talking About Birth

## Asking About Birth Date

| German | English |
|---------|----------|
| Wann bist du geboren? | When were you born? |
| Wann sind Sie geboren? | When were you born? (formal) |

---

## Answering

| German | English |
|---------|----------|
| Ich bin (im Jahr) 1990 geboren. | I was born in (the year) 1990. |

---

# Reading Years in German

1990 → neunzehnhundertneunzig  
2001 → zweitausendeins  
1985 → neunzehnhundertfünfundachtzig  

✔ Years before 2000 are usually said as:
- nineteen hundred ninety → neunzehnhundertneunzig

✔ Years after 2000:
- zweitausend...

# Sentence Pattern Reminder

Subject + sein (conjugated) + geboren

Examples:
- Ich bin 1990 geboren.
- Er ist 2005 geboren.
- Wir sind 1998 geboren.
