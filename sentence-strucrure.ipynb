# German Sentence Structure

## Basic Rules

- ✔ **The subject conjugates the verb.**

- ✔ **The verb is mostly in the second position** in a normal statement.

---

## Question Types

### 1️⃣ W-Questions (W-Fragen)

Questions beginning with **W-words** (e.g., *Wo? Woher? Wie? Was? Warum?*)  
→ The **verb is in the second position**.

**Structure:**

W-word + Verb + Subject + ...

**Examples:**

- Wo wohnst du?  
- Wie heißt du?  
- Woher kommen Sie?

---

### 2️⃣ Yes/No Questions (Ja/Nein-Fragen)

Questions answered with **"ja" (yes)** or **"nein" (no)**  
→ The **verb is in the first position**.

**Structure:**

Verb + Subject + ...

**Examples:**

- Bist du verheiratet?  
- Haben Sie Kinder?  
- Kommst du aus Deutschland?

---

## Summary

- Statements → Verb in **second position**
- W-Questions → Verb in **second position**
- Yes/No Questions → Verb in **first position**
