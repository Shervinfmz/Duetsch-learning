# German Phrases for Greetings, Polite Expressions, and Questions

This list includes common German phrases for greetings, polite expressions, and questions with their English translations.

## Greetings

| **German**                                 | **English**                               |
|--------------------------------------------|-------------------------------------------|
| **Hallo!**                                 | Hello!                                    |
| **Guten Tag!**                             | Good day!                                |
| **Guten Morgen!**                          | Good morning!                            |
| **Guten Abend!**                           | Good evening!                            |
| **Gute Nacht!**                            | Good night!                              |
| **Grüß Gott!**                             | Hello!                                    |
| **Grüß Sie / Grüß dich!**                  | Hello!                                    |
| **Servus!**                                | Hello!                                    |
| **Tschüs!**                                | Bye!                                      |
| **Auf Wiedersehen!**                       | See you again!                           |
| **Auf Wiederhören!**                       | Bye! (on the phone)                      |
| **Bis bald!**                              | See you soon!                            |
| **Bis später!**                            | See you later!                           |

## Polite Expressions

| **German**                                 | **English**                               |
|--------------------------------------------|-------------------------------------------|
| **Danke / Danke schön! / Vielen Dank!**    | Thanks! / Thank you very much!            |
| **Bitte / Bitte schön!**                   | You are welcome                           |
| **Bitte!**                                 | Please                                    |
| **Entschuldigen Sie, bitte!**              | Excuse me please (formal)                 |
| **Entschuldigung, bitte!**                 | Excuse me please (informal)               |
| **Es tut mir leid**                        | I am sorry                                |
| **Wie bitte? / Noch einmal, bitte!**       | I beg your pardon! / Once again please!  |
| **Bitte wiederholen Sie!**                 | Please repeat!                           |

## Questions

| **German**                                 | **English**                               |
|--------------------------------------------|-------------------------------------------|
| **Sprechen Sie Deutsch / Englisch?**       | Do you speak German / English?           |
| **Ich spreche etwas / kein Deutsch.**      | I speak a bit of / I don't speak German.  |
| **Ich verstehe nicht.**                    | I don't understand.                       |
| **Ich weiß nicht. / Keine Ahnung!**        | I don't know. / No idea!                 |
