---

## Vocabulary Note

- **das Gegenteil** = the opposite (singular)  
- **die Gegenteile** = the opposites (plural)

---

## Example Sentences

- Das Haus ist groß, nicht klein.  
- Der Film ist gut, nicht schlecht.  
- Das Wasser ist kalt, nicht warm.  
- Das Auto ist teuer, nicht billig.
