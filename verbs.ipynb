# German Verbs (A1 Overview)

---

## What is a Verb?

- The verb describes an action.
- The structure of a sentence depends on the verb.
- Every verb has an **infinitive form**.
- The infinitive consists of:
  - **Verb stem (Verbstamm)**
  - **Ending (Endung)**
- When using a verb in a sentence, it must be **conjugated according to the subject**.
- There are two types of verbs:
  - **regelmäßige Verben** (regular verbs)
  - **unregelmäßige Verben** (irregular verbs)

# Types of Verbs

## 1️⃣ Regelmäßige Verben (Regular Verbs)

Also known as **schwache Verben** (weak verbs).

✔ The **verb stem does not change**.

### Example

lernen → du lernst  
machen → du machst  
wohnen → du wohnst  

---

## 2️⃣ Unregelmäßige Verben (Irregular Verbs)

Also known as **starke Verben** (strong verbs).

⚠ The **verb stem changes in certain cases** (usually with "du" and "er/sie/es").

### Example

sprechen → du sprichst  
sehen → du siehst  
fahren → du fährst

# Verb Position Reminder

✔ The verb is usually in the **second position** in statements.

Example:

- Ich komme aus China.

✔ In yes/no questions, the verb is in the **first position**.

Example:

- Kommst du aus Japan?

# Verb Conjugation in Present Tense

Infinitive = **Verb stem (Verbstamm) + ending (Endung)**

✔ The **verb stem does NOT change** (for regular verbs).  
✔ The **ending depends on person and number** (singular/plural).

---

## Present Tense Endings (Regular Verbs)

| Person | Singular | Plural |
|--------|----------|--------|
| ich | Verbstamm + **e** | wir | Verbstamm + **en** |
| du | Verbstamm + **st** | ihr | Verbstamm + **t** |
| er/sie/es | Verbstamm + **t** | sie | Verbstamm + **en** |
| Sie (formal) | Verbstamm + **en** | Sie (formal) | Verbstamm + **en** |

# Special Case 1  
## Verb Stem Ends in -s, -ß, -z

If the **verb stem ends in -s, -ß or -z**,  
the ending **-st becomes -t** (because of pronunciation).

### Example: reisen (to travel)

| Singular | Plural |
|----------|--------|
| ich reise | wir reisen |
| du reist | ihr reist |
| er/sie/es reist | sie reisen |
| Sie reisen | Sie reisen |

| Sie tanzen | Sie tanzen |

# Special Case 2  
## Verb Stem Ends in -d or -t

If the **verb stem ends in -d or -t**,  
an extra **-e-** is added before the ending.

- -st → becomes **-est**
- -t → becomes **-et**

### Example: arbeiten (to work)

| Singular | Plural |
|----------|--------|
| ich arbeite | wir arbeiten |
| du arbeitest | ihr arbeitet |
| er/sie/es arbeitet | sie arbeiten |
| Sie arbeiten | Sie arbeiten |

# Summary

Regular verbs:
- Stem does NOT change
- Only the ending changes
- Watch for:
  - -s / -ß / -z endings
  - -d / -t endings
