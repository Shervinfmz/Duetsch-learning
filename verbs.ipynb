# German Verbs (A1 Overview)

---

## What is a Verb?

- The verb describes an action.
- The structure of a sentence depends on the verb.
- Every verb has an **infinitive form**.
- The infinitive consists of:
  - **Verb stem (Verbstamm)**
  - **Ending (Endung)**
- When using a verb in a sentence, it must be **conjugated according to the subject**.
- There are two types of verbs:
  - **regelmäßige Verben** (regular verbs)
  - **unregelmäßige Verben** (irregular verbs)

# Types of Verbs

## 1️⃣ Regelmäßige Verben (Regular Verbs)

Also known as **schwache Verben** (weak verbs).

✔ The **verb stem does not change**.

### Example

lernen → du lernst  
machen → du machst  
wohnen → du wohnst  

---

## 2️⃣ Unregelmäßige Verben (Irregular Verbs)

Also known as **starke Verben** (strong verbs).

⚠ The **verb stem changes in certain cases** (usually with "du" and "er/sie/es").

### Example

sprechen → du sprichst  
sehen → du siehst  
fahren → du fährst

# Verb Position Reminder

✔ The verb is usually in the **second position** in statements.

Example:

- Ich komme aus China.

✔ In yes/no questions, the verb is in the **first position**.

Example:

- Kommst du aus Japan?
